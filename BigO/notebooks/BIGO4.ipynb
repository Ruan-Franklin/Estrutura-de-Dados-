# Notação BIG O

## Exemplos BIg O

### Constant - O(1)

```python
lista = [1, 2, 3, 4, 5]
def constant(n):
    print(n[1])
    
constant(lista)
    
```

## Linear

```python
def linear(n):
    for i in n:
        print(i)
        
linear(lista)
```

## Quadrática

```python
def quadratica(n):
    for i in n:
        for j in n:
            print(i, j)

quadratica(lista)
            
```
